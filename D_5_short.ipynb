```python
import itertools
import sympy as S
```

```python
s_0, s_1, s_2, s_3, t = S.symbols('s_0 s_1 s_2 s_3 t')
xs=(s_0, s_1,s_2,s_3,t)
p,q,b,c = S.symbols('p q b c')
metric=S.zeros(5)
for i,j in ((0,3),(3,0),(1,2),(2,1)):
    metric[i,j]=1
metric[4,4]=-4
inverse=metric.inv()
unit=S.Rational(1,2)*s_0**2*s_3+s_0*s_1*s_2-2*s_0*t**2
```

```python
# 16 возможных мономов, независимых от s_0
monoms=(
    s_1**3, s_1**2*s_2*s_3, s_1**2*s_3**3, s_1*s_2**3,
    s_1*s_2**2*s_3**2,s_1*s_2*s_3**4,s_1*s_3**6,s_1*s_3*t**2,
    s_2**4*s_3,s_2**3*s_3**3,s_2**2*s_3**5,s_2**2*t**2,
    s_2*s_3**7,s_2*s_3**2*t**2,s_3**9,s_3**4*t**2,
)
coeff=S.symbols('c0:16')
F_general=unit+sum(a*m for a,m in zip(coeff,monoms))
```

```python
def multiplication_matrices(F):
    d3=[[[S.diff(F,xs[i],xs[j],xs[k]) for k in range(5)]
         for j in range(5)] for i in range(5)]
    return [S.Matrix(5,5,lambda r,s:sum(inverse[r,k]*d3[i][s][k]
                                            for k in range(5)))
            for i in range(5)]
```

```python
def wdvv_constraints(L):
    equations=set()
    for i,j in itertools.combinations(range(1,5),2):
        for expression in L[i]*L[j]-L[j]*L[i]:
            if expression != 0:
                for _,coefficient in S.Poly(expression,s_1,s_2,s_3,t).terms():
                    equations.add(S.expand(coefficient))
    return equations
```

```python
L=multiplication_matrices(F_general)
assert L[0]==S.eye(5)
equations=wdvv_constraints(L)
print('Coefficient equations from all six WDVV commutators:',len(equations))
```

```python
# Branch e1^2|0 = 0, i.e. c0 = 0.
branch={a:S.Integer(0) for a in coeff}
branch.update({coeff[3]:b,coeff[11]:c,
               coeff[8]:(c**2+6*b*c)/12})
assert all(S.expand(E.subs(branch))==0 for E in equations)
F_degenerate=S.expand(F_general.subs(branch))
assert S.diff(F_degenerate,s_3,2)==0
print('лист a=0:',F_degenerate)
print('e_3^2 = 0 тождественно, so never semisimple.')
```
